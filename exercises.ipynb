```python
#exo 1 
class Complex : 
    def __init__(self, p_reel, p_complex):
        #constructor : how to create an object
        self.number = [p_reel, p_complex]
        
    def addition(self, added_numb):
        result = [self.number[0]+added_numb[0], self.number[1]+added_numb[1]]
        return f"{result[0]}+{result[1]}i"
                  
    def substraction(self, sub_numb):
        result = [self.number[0]-sub_numb[0], self.number[1]-sub_numb[1]]
        return f"{result[0]}+{result[1]}i"
```

```python
test = Complex(5,5)
```

```python
test.addition([2,9])
```

## Problem 2

```python
#exo 2
class Person :
    def __init__(self, name, age, gender):
        self.name = str(name)
        self.age = int(age)
        self.gender = str(gender)

    def more_than(self, other):
        if self.age > other.age :
            return True
        else:
            return False
```

```python
Angelique = Person('Angelique', 49, 'F')
Oscar = Person ('Oscar', 19, 'H')
Angelique.more_than(Oscar)
```

```python
Camille = Person('Camille', 18, 'F')
Camille.more_than(Oscar)
```

## Problem 4

```python
#exo4
class Vector :
    def __init__(self, x, y , z):
        self.x = float(x)
        self.y = float(y)
        self.z = float(z)

    def dot_prod(self, v2):
        result = self.x * v2.x + self.y * v2.y + self.z * v2.z
        return result
```

```python
v1 = Vector(2,4,6)
v2 = Vector(1,3,5)
v1.dot_prod(v2)
```
